```python
import numpy as np
import matplotlib.pyplot as plt

tau=2
V0=12
Vtau=.368*V0

t=np.arange(0,100,.01)
V=V0*np.e**(-t/tau)

plt.figure(figsize=(8,4))
plt.xlim(0,11)
plt.ylim(0,1.1*V0)
plt.hlines(V0,0,20,linestyle='--',color='gray',lw=1)
plt.hlines(Vtau,0,tau,linestyle='--',color='gray',lw=1)
plt.vlines(tau,0,Vtau,linestyle='--',color='gray',lw=1)
plt.xlabel('Time',fontsize=12)
plt.ylabel('Voltage',fontsize=12)
plt.yticks([0,Vtau,V0],['0','   0.368'+'$V_0$','$V_0$'])
plt.xticks([0,tau],['0','$\\tau$'])
plt.plot(t,V,color='black')
```

```python
import numpy as np
import matplotlib.pyplot as plt

tau=2
V0=12
Vtau=.632*V0

t=np.arange(0,100,.01)
V=V0*(1-np.e**(-t/tau))

plt.figure(figsize=(8,4))
plt.xlim(0,11)
plt.ylim(0,1.1*V0)
plt.hlines(V0,0,20,linestyle='--',color='gray',lw=1)
plt.hlines(Vtau,0,tau,linestyle='--',color='gray',lw=1)
plt.vlines(tau,0,Vtau,linestyle='--',color='gray',lw=1)

plt.xlabel('Time',fontsize=12)
plt.ylabel('Voltage',fontsize=12)
plt.yticks([0,Vtau,V0],['0','   0.632'+'$V_0$','$V_0$'])
plt.xticks([0,tau],['0','$\\tau$'])
plt.plot(t,V,color='black')
```
